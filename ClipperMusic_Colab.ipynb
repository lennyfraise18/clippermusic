## Cellule 2 — Tes clés API

Elles ne sont **pas enregistrées** : il faut les recoller à chaque nouvelle session Colab.
Tu les retrouves dans le fichier `.env` de ton dossier sur ton PC.

```
import os
from getpass import getpass

# getpass masque la saisie : les clés n'apparaissent pas à l'écran
# et ne restent pas dans le notebook si tu le partages.
os.environ["PEXELS_API_KEY"] = getpass("Clé Pexels     : ")
os.environ["PIXABAY_API_KEY"] = getpass("Clé Pixabay    : ")
os.environ["JAMENDO_CLIENT_ID"] = getpass("Client ID Jamendo : ")

print("\n✅ Clés enregistrées pour cette session. Passe à la cellule 3.")
```
